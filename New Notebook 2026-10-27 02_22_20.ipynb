```sql
--Question 1
ALTER TABLE employees ADD COLUMN salary STRING;
insert into employees values
('1', 'Alice', 'HR', '5000'),
('2', 'Bob', 'IT', NULL),
('3', 'Charlie', NULL, '7000'),
('4', 'Dana', 'Finance', NULL);
select* from employees; 
--1.1. 
select 
employee_id,
employee_name,
department,
salary,
CASE
    WHEN salary IS NULL THEN 0
    ELSE salary
    END AS salary_with_default
from employees;
--1.2.
select
employee_id,
employee_name,
CASE 
    WHEN department IS NULL THEN 'Not Assigned'
    ELSE department
    END AS department_name
from employees;
--Question 2
create or replace table orders (
    order_id STRING,
    customer_id STRING,
    delivery_date DATE
);
insert into orders values
('101', '201', DATE'2024-12-01'),
('102', '202', NULL),
('103', NULL, DATE'2024-12-03');
SELECT* FROM orders;
--2.1.
select
order_id,
customer_id
from orders
where customer_id IS NULL;
--2.2. 
select
order_id,
customer_id,
CASE 
   WHEN delivery_date IS NULL THEN 'Pending'
   ELSE CAST(delivery_date AS STRING)
   END AS delivery_status
from orders;
--Question 3
create or replace table students (
    student_id STRING,
    name STRING,
    grade STRING
);
insert into students values
('1','Ethan', '85'),
('2','Maya', NULL),
('3','Olivia', '90');
select* from students;
--3.1.
select
student_id,
name,
CASE 
    WHEN grade IS NULL THEN 0
    ELSE grade 
    END AS final_grade
from students;
--3.2
select COUNT(grade) AS not_graded_count
from students;
--Question 4
create or replace table products (
product_id STRING,
name STRING,
price STRING,
discount STRING
);
insert into products values 
('501', 'Keyboard', '25', NULL),
('502', 'Mouse', '15', '5'),
('503', 'Monitor', '100', NULL);
select* from products; 
--4.1.
select
product_id,
name,
price,
discount,
CASE
    WHEN discount IS NULL THEN 0
    ELSE discount
    END AS final_price
FROM products;
--Question 5
create or replace table customers (
customer_id STRING,
name STRING,
email STRING
);
insert into customers values
('1', 'Linda', NULL),
('2', 'Joseph', 'joseph@mail.com'),
('3', 'Nia', NULL);
select* from customers;
--5.1.
select
COUNT(*) AS missing_email_count
from customers
WHERE email IS NULL;
--5.2. 
select 
customer_id,
name,
CASE 
    WHEN email IS NULL THEN 'No Email'
    ELSE email
    END AS email_display
from customers;
--Question 6
create or replace table payments (
payment_id STRING,
method STRING,
status STRING
);
insert into payments values
('301', 'Credit', NULL),
('302', 'Paypal', 'Success'),
('303', NULL, 'Failed');
select* from payments;
--6.1. 
select 
payment_id,
CASE 
    WHEN method IS NULL THEN 'Unknown'
    ELSE method
    END AS method_display,
status
from payments;
--Question 7
create or replace table inventory (
    item_id STRING,
    item_name STRING,
    quantity STRING
);
insert into inventory values 
('1', 'Pen', NULL),
('2', 'Notebook', '150'),
('3', 'Eraser', NULL);
select* from inventory;
--7.1. 
select
item_id,
item_name,
CASE 
    WHEN quantity IS NULL THEN 0
    ELSE quantity
    END AS quantity_checked
from inventory;
--Question 8
create or replace table employees_extra (
emp_id STRING,
bonus STRING,
commission STRING
);
insert into employees_extra values
('1', NULL, '300'),
('2', '100', NULL),
('3', NULL, NULL);
select* from employees_extra; 
--8.1. 
select
emp_id
from employees_extra
WHERE bonus IS NOT NULL OR commission IS NOT NULL;
--Question 9
create or replace table classes (
class_id STRING,
subject STRING,
room STRING
);
INSERT into classes values
('11', 'Math', NULL),
('12', 'Science', 'Lab A'),
('13', 'English', NULL);
Select* from classes;
--9.1. ,
select 
count(*) as no_room_count
from classes
where room is null;
--Question 10
create or replace table attendance (
    student_id STRING,
    date DATE,
    status STRING
);
INSERT INTO attendance values
('1', DATE'2025-04-01', NULL),
('2', DATE'2025-04-01', 'Present'),
('3', DATE'2025-04-01', 'Absent');
select* from attendance; 
--10.1.
select
student_id,
date,
CASE
    WHEN status IS NULL THEN 'Not Marked'
    ELSE status
    END AS attendance_status
from attendance;
--Question 11
create or replace table bank_accounts  (
    account_id STRING,
    account_type STRING,
    balance STRING
);
insert into bank_accounts values 
('A1', 'Savings', NULL),
('A2', 'Current', '5000'),
('A3', NULL, '2000');
select* from bank_accounts;
--11.1.
select 
account_id,
CASE
    WHEN account_type IS NULL THEN 'Unknown'
    ELSE account_type
    END AS type_display,
    CASE
    WHEN balance IS NULL THEN 0
    ELSE balance 
    END AS balance_checked
from bank_accounts;
--Question 12
create or replace table projects (
    project_id STRING,
    title STRING,
    start_date DATE,
    end_date DATE
);
insert into projects values 
('1', 'Website Revamp', DATE'2025-01-10', NULL),
('2', 'Mobile App', NULL, DATE'2025-06-01'),
('3', 'Data Migration', NULL, NULL);
select* from projects; 
--12.1.
select
project_id,
title,
CASE
    WHEN start_date IS NULL THEN 'TBD'
    ELSE CAST(start_date AS STRING)
    END AS start_display
from projects;
--Question 13 
create or replace table reviews (
    review_id STRING,
    product_id STRING,
    comment STRING,
    rating STRING
);
insert into reviews values 
('1', '501', 'Great product', '4'),
('2', '502', NULL, NULL),
('3', '503', 'Works fine', '3');
select* from reviews;
--13.1.
select
review_id,
product_id,
CASE 
    WHEN comment IS NULL THEN 'No Comment'
    ELSE comment
    END AS comment_display,
CASE
    WHEN rating IS NULL THEN 0
    ELSE rating 
    END AS rating_display
from reviews;
--Question 14
create or replace table suppliers (
    supplier_id STRING,
    name STRING,
    phone STRING,
    alt_phone STRING
);
insert into suppliers values 
('1', 'Global Goods', NULL, '123456789'),
('2', 'Best Supplies', '987654321', NULL),
('3', 'ValueSource', NULL, NULL);
select* from suppliers;
--14.1.
select
supplier_id,
name,
COALESCE(phone, alt_phone, 'No Contact') AS contact_number
from suppliers;
--Question 15
create or replace table user_settings (
    user_id STRING,
    theme STRING,
    language STRING,
    timezone STRING
);
insert into user_settings values
('1', NULL, 'English', NULL),
('2', 'Dark', NULL, 'UTC+1'),
('3', NULL, NULL, NULL);
select* from user_settings;
--15.1.
select
user_id,
CASE 
    WHEN theme IS NULL THEN 'Light'
    ELSE theme
    END AS theme_set,

CASE
    WHEN language IS NULL THEN 'English'
    ELSE language 
    END AS language_set,

CASE 
    WHEN timezone IS NULL THEN 'UTC'
    ELSE timezone
    END AS timezone_set
from user_settings;
--Question 16
create table maintenance (
    record_id STRING,
    machine_id STRING,
    issue STRING,
    technician STRING
);
insert into maintenance values
('1', 'M101', 'Overheating', NULL),
('2', 'M102', NULL, NULL),
('3', 'M103', 'Jammed', 'Alex');
select* from maintenance; 
--16.1. 
select
record_id,
machine_id,
CASE 
    WHEN issue IS NULL THEN 'Unknown Issue'
    ELSE issue
    END AS issue_log,

CASE 
    WHEN technician IS NULL THEN 'Not Assigned'
    ELSE technician
    END AS technician_name
from maintenance;
```
